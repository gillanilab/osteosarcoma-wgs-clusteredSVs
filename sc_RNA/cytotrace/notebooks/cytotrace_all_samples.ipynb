```python
import cytotrace2_py as cytotrace2
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
from scipy.stats import pearsonr
```

```python
# run cytotrace on all samples, concatenate results in csv

samples = {
    "SJOS001108_M1": "../../data/st_jude/SJOS001108_M1.h5ad", # TP53 upstream cluster
    "SJOS016015_D1": "../../data/st_jude/SJOS016015_D1.h5ad", # TP53 upstream cluster
    "SJOS030597_D1": "../../data/st_jude/SJOS030597_D1.h5ad", # MYC cluster
    "SJOS012409_D1": "../../data/st_jude/SJOS012409_D1.h5ad", # no hotspot clusters
    "SJOS046149_R1": "../../data/st_jude/SJOS046149_R1.h5ad", # CCND3 cluster
    "SJOS001107_M2": "../../data/st_jude/SJOS001107_M2.h5ad", # TP53 upstream cluster, CCND3 cluster
    "SJOS063833_D2": "../../data/st_jude/SJOS063833_D2.h5ad", # MYC cluster
    "SJOS030101_D1": "../../data/st_jude/SJOS030101_D1.h5ad", # TP53 upstream cluster, CCND3 cluster
    "CCG1106_086_T1": "../../data/dfci/expression-data_scrna_dfci_tumor_expr_CCG1106_086_T1.h5ad", # TP53 upstream cluster, CCND3 cluster
    "17_439_00067_T2": "../../data/dfci/expression-data_scrna_dfci_tumor_expr_17_439_00067_T2.h5ad", # no hotspot clusters
    "CCG1106_005_T1": "../../data/dfci/expression-data_scrna_dfci_tumor_expr_CCG1106_005_T1.h5ad", # TP53 upstream cluster
    "CCG1106_075_T1": "../../data/dfci/expression-data_scrna_dfci_tumor_expr_CCG1106_075_T1.h5ad", # TP53 upstream cluster
    "CCG1106_083_T1": "../../data/dfci/expression-data_scrna_dfci_tumor_expr_CCG1106_083_T1.h5ad", # no hotspot clusters
    "CCG1106_084_T1": "../../data/dfci/expression-data_scrna_dfci_tumor_expr_CCG1106_084_T1.h5ad" # MYC cluster
}

# load gene annotations
genes = pd.read_csv("../../../bulk_RNA/data/gene_coordinates.txt", sep="\t")
genes['Chromosome/scaffold name'] = genes['Chromosome/scaffold name'].replace({'X': '23', 'Y': '24'})
genes['Chromosome/scaffold name'] = pd.to_numeric(genes['Chromosome/scaffold name'])
genes = genes[(genes['Chromosome/scaffold name'] >= 1) & (genes['Chromosome/scaffold name'] <= 24)]

# Define regions of interest
regions = {
    "MYC": (8, 99000000, 137000000),
    "CCND3": (6, 39000000, 55000000),
    "TP53_upstream": (17, 7000000, 22000000)
}

region_genes = {}
for name, (chrom, start, end) in regions.items():
    region_genes[name] = genes[
        (genes["Chromosome/scaffold name"] == chrom) &
        (genes["Gene start (bp)"] <= end) & 
        (genes["Gene end (bp)"] >= start)
    ]["Gene name"].dropna().tolist()

results_list = []

for sample_id, filepath in samples.items():
    print(f"Processing {sample_id}...")

    adata = sc.read_h5ad(filepath) # Load data
    adata = adata[adata.obs['n_genes_by_counts'] > 500].copy() # Filter cells
    
    # Save expression matrix (raw counts)for CytoTRACE
    expr_matrix = pd.DataFrame(
        adata.X.toarray().T if hasattr(adata.X, 'toarray') else adata.X.T,
        index=adata.var_names,
        columns=adata.obs_names
    )
    expr_matrix.to_csv('../outputs/cytotrace_temp.txt', sep='\t')
    
    # Normalize for region score calculation
    adata_norm = adata.copy()
    sc.pp.normalize_total(adata_norm, target_sum=1e4)
    sc.pp.log1p(adata_norm)
    
    # Calculate region scores
    region_scores = {}
    for region_name, gene_list in region_genes.items():
        genes_present = [g for g in gene_list if g in adata_norm.var_names]
        if genes_present:
            region_data = adata_norm[:, genes_present].X
            region_scores[f'{region_name}_score'] = (
                region_data.toarray() if hasattr(region_data, 'toarray') else region_data
            ).mean(axis=1)
    
    # Run CytoTRACE2
    ct_results = cytotrace2.cytotrace2(input_path='../outputs/cytotrace_temp.txt', species='human')
    
    # Combine results
    ct_results['sample_id'] = sample_id
    ct_results['cell_barcode'] = ct_results.index
    ct_results['n_genes_by_counts'] = adata.obs['n_genes_by_counts'].values
    ct_results['total_counts'] = adata.obs['total_counts'].values
    
    for score_name, scores in region_scores.items():
        ct_results[score_name] = scores
    
    results_list.append(ct_results)

# Combine results across samplesand save
results_df = pd.concat(results_list, ignore_index=True)
results_df.to_csv('../outputs/cytotrace_all_samples.csv', index=False)
print(f"Analysis complete! Results saved with {len(results_df)} cells across {len(samples)} samples")
print(f"Columns: {results_df.columns.tolist()}")
```

```python
# Association of TP53 upstream score with CytoTRACE2 score (for sample of interest)
df = pd.read_csv('../outputs/cytotrace_all_samples.csv')
df_sample = df[df['sample_id'] == 'CCG1106_075_T1']

df_sample = df_sample.dropna(subset=['CytoTRACE2_Score', 'TP53_upstream_score', 'n_genes_by_counts', 'total_counts'])

# Create categorical variable based on threshold
df_sample['TP53_upstream_amplified'] = (df_sample['TP53_upstream_score'] > 0.2).astype(int)
y = df_sample['CytoTRACE2_Score']
X = df_sample[['TP53_upstream_amplified', 'n_genes_by_counts', 'total_counts']].astype(float)

X = sm.add_constant(X)

# Fit the model
model = sm.OLS(y, X)
results = model.fit()

print(results.summary())
```

```python
# Association of MYC score with CytoTRACE2 score (for sample of interest)
df_sample = df[df['sample_id'] == 'CCG1106_084_T1']

df_sample = df_sample.dropna(subset=['CytoTRACE2_Score', 'MYC_score', 'n_genes_by_counts', 'total_counts'])

# Create categorical variable based on MYC_score threshold
df_sample['MYC_amplified'] = (df_sample['MYC_score'] > 0.35).astype(int)

y = df_sample['CytoTRACE2_Score']
X = df_sample[['MYC_amplified', 'n_genes_by_counts', 'total_counts']].astype(float)

X = sm.add_constant(X)

# Fit the model
model = sm.OLS(y, X)
results = model.fit()

print(results.summary())

print(f"\nMYC amplification status breakdown:")
print(df_sample['MYC_amplified'].value_counts())
```

```python
# plot cytotrace scores for all samples (violin plot per sample)
score_col = "CytoTRACE2_Score"
sample_col = "sample_id"

plot_df = df[[sample_col, score_col]].dropna().copy()

order = (
    plot_df.groupby(sample_col)[score_col]
    .median()
    .sort_values(ascending=False)
    .index
)

plt.figure(figsize=(0.5 * len(order) + 6, 5))
ax = sns.violinplot(
    data=plot_df,
    x=sample_col,
    y=score_col,
    order=order,
    inner="quartile",
    cut=0,
    linewidth=0.8,
    color="lightgray",
)

ax.set_xlabel("")
ax.set_ylabel("CytoTRACE2 score")
ax.tick_params(axis="x", labelrotation=90)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()
```
