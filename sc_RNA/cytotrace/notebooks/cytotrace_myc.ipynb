```python
import cytotrace2_py as cytotrace2
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
results = cytotrace2.cytotrace2(
    input_path='../data/CCG1106_084_T1_expression_matrix_raw.txt',
    annotation_path='../data/CCG1106_084_T1_phenotype.txt',
    species='human'
)

print("CytoTRACE2 completed!")
print("\nResults keys:", results.keys())
```

```python
cytotrace_scores = results['CytoTRACE2_Score'] 
phenotype_data = pd.read_csv('../data/CCG1106_084_T1_phenotype.txt', sep='\t')

# Merge scores with phenotypes
results_df = pd.DataFrame(cytotrace_scores)
results_df = results_df.merge(phenotype_data, left_index=True, right_on='cell_id')

print("\nResults preview:")
print(results_df.head())

# Compare groups
amplified_scores = results_df[results_df['phenotype'] == 'MYC_amplified']['CytoTRACE2_Score']
non_amplified_scores = results_df[results_df['phenotype'] == 'Non_amplified']['CytoTRACE2_Score']

print(f"\nMYC amplified - mean CytoTRACE2 score: {amplified_scores.mean():.3f} (n={len(amplified_scores)})")
print(f"Non-amplified - mean CytoTRACE2 score: {non_amplified_scores.mean():.3f} (n={len(non_amplified_scores)})")

# Statistical test
stat, pval = stats.mannwhitneyu(amplified_scores, non_amplified_scores, alternative='two-sided')
print(f"\nMann-Whitney U test p-value: {pval:.2e}")

def get_stars(p):
    if p < 0.001: return '***'
    elif p < 0.01: return '**'
    elif p < 0.05: return '*'
    else: return 'ns'

palette = {
    'MYC_amplified': '#ee876cdb',
    'Non_amplified': '#CCCCCC'
}

fig, ax = plt.subplots(figsize=(6, 6))
sns.violinplot(
    data=results_df,
    x='phenotype',
    y='CytoTRACE2_Score',
    ax=ax,
    palette=palette,
    inner='box',
    inner_kws=dict(
        box_width=10,
        whis_width=2,
        color='black',
    )
)
ax.grid(False)
sns.despine()

# Add significance bracket
y_max = results_df['CytoTRACE2_Score'].max() + 0.1
y_range = results_df['CytoTRACE2_Score'].max() - results_df['CytoTRACE2_Score'].min()
bar_y = y_max + y_range * 0.05 
tip_y = bar_y - y_range * 0.02

ax.plot([0, 0, 1, 1], [tip_y, bar_y, bar_y, tip_y], lw=1, color='black')
ax.text(0.5, bar_y + y_range * 0.01, get_stars(pval), ha='center', va='bottom', fontsize=14)

ax.set_title('CytoTRACE2 Scores by MYC Clustered SV event Status')
ax.set_ylabel('CytoTRACE2 Score')
ax.set_xlabel('')

plt.tight_layout()
plt.savefig(f"../outputs/cytotrace2_output_CCG1106_084_T1/cytotrace_score_by_event_violin.pdf", dpi=300)
plt.show()
```

How do CytoTRACE scores for cells from SJOS001107_M2 (homogenous event-positive population) compare? 

```python
cytotrace_scores_myc = pd.read_csv('../outputs/cytotrace2_output_CCG1106_084_T1/cytotrace2_results.txt', sep='\t')
# set first column as 'cell_id'
cytotrace_scores_myc = cytotrace_scores_myc.rename(columns={cytotrace_scores_myc.columns[0]: 'cell_id'})
phenotype_data_myc = pd.read_csv('../data/CCG1106_084_T1_phenotype.txt', sep='\t')
phenotype_data_myc.head()

results_df_myc = pd.DataFrame(cytotrace_scores_myc)
results_df_myc = cytotrace_scores_myc.merge(phenotype_data_myc, on='cell_id', how='left')

cytotrace_scores_tp53 = pd.read_csv('../outputs/cytotrace2_output_CCG1106_075_T1_TP53/cytotrace2_results.txt', sep='\t')
cytotrace_scores_tp53 = cytotrace_scores_tp53.rename(columns={cytotrace_scores_tp53.columns[0]: 'cell_id'})
phenotype_data_tp53 = pd.read_csv('../data/CCG1106_075_T1_TP53_phenotype.txt', sep='\t')

results_df_tp53 = pd.DataFrame(cytotrace_scores_tp53)
results_df_tp53 = cytotrace_scores_tp53.merge(phenotype_data_tp53, on='cell_id', how='left')

# filter combined sample results to ccnd3 sample
all_samples_cytotrace = pd.read_csv('../outputs/cytotrace_all_samples.csv')
results_df_ccnd3 = all_samples_cytotrace[all_samples_cytotrace['sample_id'] == 'SJOS001107_M2']
```

```python
# Add phenotype column to CCND3 (all cells same group)
results_df_ccnd3['phenotype'] = 'CCND3_all'

# Combine all three into one dataframe with a sample label
results_df_myc['sample']   = 'MYC'
results_df_tp53['sample']  = 'TP53'
results_df_ccnd3['sample'] = 'CCND3'

plot_df = pd.concat([results_df_myc, results_df_tp53, results_df_ccnd3], ignore_index=True)

# Define violin order: MYC split, TP53 split, CCND3 all
order = ['MYC_amplified', 'Non_amplified_MYC', 'TP53_upstream_amplified', 'Non_amplified_TP53', 'CCND3_all']

plot_df['violin_group'] = plot_df['phenotype'].copy()
plot_df.loc[plot_df['sample'] == 'MYC',  'violin_group'] = plot_df.loc[plot_df['sample'] == 'MYC',  'phenotype'].replace({'Non_amplified': 'Non_amplified_MYC'})
plot_df.loc[plot_df['sample'] == 'TP53', 'violin_group'] = plot_df.loc[plot_df['sample'] == 'TP53', 'phenotype'].replace({'Non_amplified': 'Non_amplified_TP53'})

palette = {
    'MYC_amplified':          '#ee876cdb',
    'Non_amplified_MYC':      '#CCCCCC',
    'TP53_upstream_amplified': '#ee876cdb',
    'Non_amplified_TP53':     '#CCCCCC',
    'CCND3_all':              '#ee876cdb'
}

fig, ax = plt.subplots(figsize=(8, 6))

sns.violinplot(
    data=plot_df,
    x='violin_group',
    y='CytoTRACE2_Score',
    order=order,
    ax=ax,
    palette=palette,
    inner='box',
    inner_kws=dict(
        box_width=6,
        whis_width=2,
        color='gray',
    )
)

# Sample group labels above
ax.text(0.5, 1.02, 'MYC', transform=ax.get_xaxis_transform(),
        ha='center', fontsize=9, fontweight='bold')
ax.text(2.5, 1.02, 'TP53 upstream', transform=ax.get_xaxis_transform(),
        ha='center', fontsize=9, fontweight='bold')
ax.text(4.0, 1.02, 'CCND3', transform=ax.get_xaxis_transform(),
        ha='center', fontsize=9, fontweight='bold')

ax.set_xlabel('')
ax.set_ylabel('CytoTRACE2 Score', fontsize=10)
ax.set_xticklabels(['Clustered SV', 'Non', 'Clustered SV', 'Non', 'All'], fontsize=8)
ax.grid(False)
sns.despine()

plt.tight_layout()
plt.savefig(f"../outputs/cytotrace_score_by_event_violin.pdf", dpi=300)
plt.show()
```
